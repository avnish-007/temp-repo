# Data Visulization
It is the process of converting the data into pictorial or graphical format to understand the data in better manner.

**Three ways to Visulize the data**
1. Matplotlib (libraries)
2. Seaborn (libraries)
3. PowerBI (tool)

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.arange(1,11)
y = x ** 2
```

```python
plt.plot(x,y,marker='p') # o,*,h,s,p,x
```

```python
plt.plot(x,y,color= "#FF2056")
```

```python
plt.plot(x,y,marker='p',linestyle='--') # dashed / dotted / solid
```

```python
plt.plot(x,y,marker='p',linewidth= 1)
```

```python
plt.plot(x,y,marker='p', markersize= 3)
```

```python
plt.plot(x,y,marker='p',markerfacecolor='red')
```

```python
plt.plot(x,y,marker='p',markeredgecolor='red')
```

```python
plt.plot(x, y, marker='p', color='#ff2056', linewidth=4, markersize=10, markerfacecolor='#05DF72', markeredgecolor='#155DFC')
plt.grid(True)
plt.xlabel('Values of x-axis', fontsize=12, color='black')
plt.ylabel('Values of y-axis', fontsize=12, color='black')
plt.title('Graph', fontsize=12, color='black')
plt.gca().set_facecolor('Pink')
# Annotate():
for i in range(len(x)):
    plt.annotate(f'({x[i]},{y[i]})' , (x[i],y[i]))
```

```python
x = np.arange(1,11)
y = x ** 2
z = x ** 3
```

```python
plt.plot(x, z, marker='p', color='#ff2056', linewidth=4, markersize=10, markerfacecolor='#05DF72', markeredgecolor='#155DFC')
plt.grid(True)
plt.xlabel('Values of x-axis', fontsize=12, color='black')
plt.ylabel('Values of z-axis', fontsize=12, color='black')
plt.title('Graph', fontsize=12, color='black')
plt.gca().set_facecolor('Yellow')
# Annotate():
for i in range(len(x)):
    plt.annotate(f'({x[i]},{z[i]})' , (x[i],z[i]))
```
