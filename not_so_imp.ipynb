```python
# Import NumPy library
import numpy as np
```

```python
# Check NumPy version
np.__version__
```

```python
# Create a NumPy array from a Python list
l = [5,10,15,20]
arr = np.array(l)
arr
```

```python
# Create a NumPy array directly from a list literal
arr2 = np.array([5,1,22,44])
arr2
```

```python
# Create an array from a tuple of mixed types (demonstrating type coercion)
b = np.array((True, False, 3, 2.5, 2-3j, 'hi'))
b
```

```python
# Convert multiple Python lists into separate NumPy arrays
lst = [1,2,3,4,5]
lst2 = [0,9,8,7,6]
arr1 = np.array(lst)
arr2 = np.array(lst2)
```
