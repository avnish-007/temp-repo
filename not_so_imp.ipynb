```python
# Boolean Filtering:
# syntax:
#       var[condition]
```

```python
ages = np.array([[35,44,23,67,87,45,23,45,23],
                 [87,56,84,34,25,54,34,65,54]])
ages
```

```python
# Filter out those ages which are greater than 60:
ages[ages > 60]
```

```python
# Filter out the ages which are less than 18
ages[ages < 18]
```

```python
# Filter out the ages between 25 to 50
ages[(ages > 25) & (ages < 50)]
```

```python
# Filtwer out the ages which are less than 18 and more than 60.
ages[(ages < 18) | (ages > 60)]
```

```python
# Filter out the ages which are in between 18 to 60
ages[(ages > 18) & (ages < 60)]
```

```python
# Indexing and Slicing on Data
# loc and iloc

# loc --> label based indexing
# 1) var.loc[index]
# 2) var.loc[[index1,index2,. . ., indexN]]
# 3) var.loc[[index1,index2,. . ., indexN],"col_name"]
# 4) var.loc[[index1,index2,. . ., indexN],["col_name1", "col_name2", . . ., "col_nameN"]]
# 5) var.loc[SV:EV:UP]
# 6) var.loc[SV:EV:UP,"col_name"]
# 7) var.loc[SV:EV:UP, ['col_name1','col_name2', . . ., 'col_nameN']]
```
