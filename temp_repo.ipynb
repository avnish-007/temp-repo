**Question 1: Basic Single Inheritance**
Create a base class `Animal` with a method `make_sound()` that prints "Some generic sound". Create a derived class `Dog` that inherits from `Animal` and adds a new method `wag_tail()` that prints "Tail wagging". Instantiate a `Dog` object and call both methods.

```python
class Animal:
    def make_sound(self) -> None:
        print("Some generic sound")

class Dog(Animal):
    def wag_tail(self):
        print("Tail wagging")

    def __repr__(self):
        return "Dog()"

# Solution
my_dog = Dog()
my_dog.make_sound()  # Inherited method
my_dog.wag_tail()    # Dog's own method

class Cat(Animal):
    def purr(self):
        print("Purring...")

    def __repr__(self):
        return "Cat()"

my_cat = Cat()
my_cat.make_sound()  # Inherited
my_cat.purr()        # Cat's own method
```

**Question 2: Using `super()` and Initialization**
Create a base class `Person` that takes `name` and `age` as arguments in its constructor. Create a derived class `Student` that takes `name`, `age`, and `student_id`. Use the `super()` function inside the `Student` constructor to initialize `name` and `age` from the parent class.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, student_id):
        # Call the parent class constructor
        super().__init__(name, age)
        self.student_id = student_id

    def __str__(self):
        return f"{self.name} (ID: {self.student_id})"

# Solution
student = Student("Alice", 20, "S12345")
print(student)  # Uses __str__
print(f"Age: {student.age}")
```

The following questions cover polymorphism in Python OOP.

### Q1: What is polymorphism in Python OOP, and what are its primary forms?

**Answer:**
The word **polymorphism** originates from Greek, meaning *"many forms"* (`poly` = many, `morph` = form). In Object-Oriented Programming, polymorphism is the ability of different objects to respond to the same interface or method call in their own specific ways.

In Python, polymorphism manifests primarily through:
1. **Duck Typing:** Dynamic typing where behavior determines compatibility ("If it walks like a duck...").
2. **Method Overriding (Subtype Polymorphism):** A subclass providing its own implementation of a method defined in a superclass.
3. **Operator Overloading:** Defining special dunder methods (e.g., `__add__`, `__mul__`) so standard operators work with custom objects.
4. **Built-in Polymorphic Functions:** Functions like `len()` or `+` behaving differently depending on argument types.

```python
# Example: Built-in Polymorphic behavior
print("Polymorphism with len():")
print(len("Hello World"))     # Length of a string (characters)
print(len([10, 20, 30, 40]))  # Length of a list (elements)
print(len({"a": 1, "b": 2}))   # Length of a dict (keys)

print("\nPolymorphism with '+' operator:")
print(5 + 10)                  # Arithmetic addition
print("Python " + "Rocks")     # String concatenation
print([1, 2] + [3, 4])         # List merging
```

### Q2: What is "Duck Typing" in Python, and how does it relate to polymorphism?

**Answer:**
**Duck Typing** is a programming philosophy derived from the phrase:
> *"If it walks like a duck and quacks like a duck, it's a duck."*

In Python, an object's suitability for a task is determined by the presence of specific methods and properties, rather than its explicit class inheritance. Duck typing enables polymorphism without requiring inheritance from a common base class.

```python
class PDFDocument:
    def display(self):
        return "Rendering PDF document."

class WordDocument:
    def display(self):
        return "Rendering Word document (.docx)."

class ImageViewer:
    def display(self):
        return "Displaying JPG image."

# Polymorphic function relying on duck typing
def render_document(doc):
    print(doc.display())

# Notice that none of these share an inheritance tree, yet they are treated polymorphically
docs = [PDFDocument(), WordDocument(), ImageViewer()]
for doc in docs:
    render_document(doc)
```

### Q3: How does Method Overriding implement polymorphism in inheritance hierarchies?

**Answer:**
**Method Overriding** occurs when a subclass defines a method with the exact same name as a method in its parent class. When the method is invoked on an instance of the subclass, Python resolves to the subclass implementation rather than the parent's, enabling polymorphic method execution across class hierarchies.

```python
class Animal:
    def __init__(self, name: str):
        self.name = name
        
    def speak(self):
        raise NotImplementedError("Subclasses must implement abstract method")

class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"

class Cat(Animal):
    def speak(self):
        return f"{self.name} says Meow!"

class Cow(Animal):
    def speak(self):
        return f"{self.name} says Moo!"

animals = [Dog("Buddy"), Cat("Whiskers"), Cow("Bessie")]

for animal in animals:
    # Common interface, specific subtype behaviors
    print(animal.speak())
```

### Q4: Does Python support traditional Method Overloading (compile-time polymorphism)? If not, how can it be achieved?

**Answer:**
**No**, Python does not natively support compile-time method overloading (defining multiple methods with the same name but different parameter signatures in the same scope). Defining a second method with the same name simply overwrites the previous definition.

However, method overloading can be achieved via:
1. **Default/Optional arguments (`*args`, `**kwargs`, `None`)**
2. **The `functools.singledispatch` / `singledispatchmethod` decorator** (from Python's standard library)

```python
from functools import singledispatchmethod

class Formatter:
    @singledispatchmethod
    def format_data(self, data):
        raise NotImplementedError(f"Unsupported type: {type(data)}")

    @format_data.register(int)
    def _(self, data: int):
        return f"Integer value: {data:,}"

    @format_data.register(str)
    def _(self, data: str):
        return f"String value (uppercase): {data.upper()}"

    @format_data.register(list)
    def _(self, data: list):
        return f"List with {len(data)} items: {', '.join(map(str, data))}"

formatter = Formatter()
print(formatter.format_data(1000000))
print(formatter.format_data("python programming"))
print(formatter.format_data([10, 20, 30]))
```

### Q5: How do Abstract Base Classes (ABCs) enforce a polymorphic interface in Python?

**Answer:**
Python provides the `abc` module to define **Abstract Base Classes (ABCs)** and the `@abstractmethod` decorator. An ABC defines a formal polymorphic contract that all concrete subclasses **must** implement. If a subclass fails to implement any abstract method, Python prevents instantiation at runtime, raising a `TypeError`.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self) -> float:
        """Calculate and return the shape's area."""
        pass

    @abstractmethod
    def perimeter(self) -> float:
        """Calculate and return the shape's perimeter."""
        pass

class Circle(Shape):
    def __init__(self, radius: float):
        self.radius = radius

    def area(self) -> float:
        return math.pi * (self.radius ** 2)

    def perimeter(self) -> float:
        return 2 * math.pi * self.radius

class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

    def perimeter(self) -> float:
        return 2 * (self.width + self.height)

# Unified interface across diverse shapes
class Triangle(Shape):
    def __init__(self, base: float, height: float, side_a: float, side_b: float, side_c: float):
        self.base = base
        self.height = height
        self.sides = (side_a, side_b, side_c)

    def area(self) -> float:
        return 0.5 * self.base * self.height

    def perimeter(self) -> float:
        return sum(self.sides)

shapes: list[Shape] = [Circle(5), Rectangle(4, 6), Triangle(6, 4, 5, 5, 6)]
for s in shapes:
    print(f"{s.__class__.__name__} -> Area: {s.area():.2f}, Perimeter: {s.perimeter():.2f}")
```

### Q6: What is Operator Overloading, and how does it represent polymorphism?

**Answer:**
**Operator Overloading** is a form of polymorphism where standard operators (such as `+`, `-`, `*`, `==`, `<`, etc.) take on customized meanings when applied to user-defined class instances.

In Python, this is achieved by overriding special **dunder (double underscore)** methods, such as:
- `__add__` for `+`
- `__sub__` for `-`
- `__eq__` for `==`
- `__str__` / `__repr__` for string conversion

```python
class Vector2D:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def __add__(self, other):
        if isinstance(other, Vector2D):
            return Vector2D(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __mul__(self, scalar: float):
        if isinstance(scalar, (int, float)):
            return Vector2D(self.x * scalar, self.y * scalar)
        return NotImplemented

    def __eq__(self, other):
        if isinstance(other, Vector2D):
            return self.x == other.x and self.y == other.y
        return False

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"

v1 = Vector2D(2, 3)
v2 = Vector2D(5, 7)

print("v1 + v2 =", v1 + v2)
print("v1 * 3 =", v1 * 3)
print("v1 == Vector2D(2, 3):", v1 == Vector2D(2, 3))
```

### Q7: What role does Method Resolution Order (MRO) play in polymorphic method calls in Multiple Inheritance?

**Answer:**
When a class inherits from multiple parents and calls an overridden method, Python must deterministically choose which method implementation to execute. Python uses the **C3 Linearization algorithm** to construct the **Method Resolution Order (MRO)**.

The MRO defines the exact search order for attribute and method lookups. Using `super()` within methods executes polymorphic calls cooperative throughout the hierarchy according to the MRO.

```python
class Base:
    def action(self):
        print("Base action")

class MixinA(Base):
    def action(self):
        print("MixinA action")
        super().action()

class MixinB(Base):
    def action(self):
        print("MixinB action")
        super().action()

class Child(MixinA, MixinB):
    def action(self):
        print("Child action")
        super().action()

c = Child()
c.action()

print("\nMethod Resolution Order (MRO):")
for cls in Child.__mro__:
    print(" ->", cls.__name__)
```

### Q8: What is static typing polymorphism using `typing.Protocol` (Structural Subtyping)?

**Answer:**
Introduced in **PEP 544 (Python 3.8+)**, `typing.Protocol` provides **structural subtyping** (often called *static duck typing*). Instead of nominal inheritance (explicitly inheriting from a base class), any class that implements the methods specified in a `Protocol` is statically considered a subtype by static type checkers (like `mypy` or IDEs) while preserving complete runtime flexibility.

```python
from typing import Protocol

class Greeter(Protocol):
    def greet(self) -> str:
        ...

class EnglishSpeaker:
    def greet(self) -> str:
        return "Hello!"

class SpanishSpeaker:
    def greet(self) -> str:
        return "¡Hola!"

# Neither EnglishSpeaker nor SpanishSpeaker inherits from Greeter,
# yet both satisfy the Greeter protocol!
def welcome(person: Greeter) -> None:
    print(person.greet())

welcome(EnglishSpeaker())
welcome(SpanishSpeaker())

class FrenchSpeaker:
    def greet(self) -> str:
        return "Bonjour!"

welcome(FrenchSpeaker())
```

### Q9: How can polymorphism be applied in the Strategy Design Pattern to replace conditional logic (`if-elif-else`)?

**Answer:**
One of the greatest benefits of polymorphism is eliminating fragile chains of `if-elif-else` checks. In the **Strategy Pattern**, interchangeable algorithms/behaviors are encapsulated within separate classes conforming to a common interface, allowing the runtime context to switch behaviors dynamically without modifying existing client code (complying with the Open/Closed Principle).

```python
from abc import ABC, abstractmethod

class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount: float) -> str:
        pass

class CreditCardPayment(PaymentStrategy):
    def __init__(self, card_number: str):
        self.card_number = card_number

    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} using Credit Card ending in {self.card_number[-4:]}"

class PayPalPayment(PaymentStrategy):
    def __init__(self, email: str):
        self.email = email

    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} using PayPal account ({self.email})"

class CryptoPayment(PaymentStrategy):
    def __init__(self, wallet_address: str):
        self.wallet = wallet_address

    def pay(self, amount: float) -> str:
        return f"Paid ${amount:.2f} using Crypto Wallet {self.wallet[:6]}..."

# Order checkout context executes polymorphically
class Order:
    def __init__(self, total_amount: float, payment_method: PaymentStrategy):
        self.total_amount = total_amount
        self.payment_method = payment_method

    def checkout(self):
        print(self.payment_method.pay(self.total_amount))

Order(120.50, CreditCardPayment("4111222233334567")).checkout()
Order(75.00, PayPalPayment("buyer@example.com")).checkout()
Order(250.00, CryptoPayment("0x71C...8932")).checkout()
```

### Q10: What is the Liskov Substitution Principle (LSP), and why is it essential for safe polymorphism?

**Answer:**
The **Liskov Substitution Principle (LSP)** is the 'L' in SOLID design principles. It states:
> *"Objects of a superclass should be replaceable with objects of its subclasses without breaking the application or altering the desirable properties of the program."*

When implementing polymorphism via inheritance:
1. **Preconditions cannot be strengthened in a subclass** (a subclass must not demand stricter input arguments than the parent).
2. **Postconditions cannot be weakened in a subclass** (a subclass must return what the caller expects).
3. **Invariants must be preserved** (class rules must remain consistent).

Violating LSP leads to unexpected runtime bugs, requiring brittle `isinstance` checks that undermine the entire purpose of polymorphism.

```python
# Example of LSP compliance
class NotificationSender(ABC):
    @abstractmethod
    def send(self, recipient: str, message: str) -> bool:
        """Must return True if successful, False otherwise."""
        pass

class EmailSender(NotificationSender):
    def send(self, recipient: str, message: str) -> bool:
        print(f"Email sent to {recipient}: {message}")
        return True

class SMSSender(NotificationSender):
    def send(self, recipient: str, message: str) -> bool:
        print(f"SMS sent to {recipient}: {message}")
        return True

# Consumer relies on LSP guarantee that any sender works interchangeably
def notify_all(senders: list[NotificationSender], user: str, msg: str):
    for s in senders:
        success = s.send(user, msg)
        assert isinstance(success, bool), "LSP violation: return type mismatch!"

notify_all([EmailSender(), SMSSender()], "alice@test.com", "Your order has shipped!")
```

## Project 1: Student Exam Score Analyzer

### Objective
You are given exam scores for 5 students across 4 subjects (Math, Science, English, History).

### Tasks
1. Create a 2D array representing scores (out of 100) for 5 students across 4 subjects.
2. Calculate each student's average score across all subjects.
3. Find the subject with the highest overall class average.
4. Identify students who scored 80 or above in *every* subject.
5. Standardize (Min-Max normalize) the scores to a range of $[0, 1]$.

```python
import numpy as np

# Step 1: Create scores matrix (5 students x 4 subjects)
np.random.seed(42)
scores = np.array([
    [85, 78, 92, 88],
    [70, 65, 80, 75],
    [90, 95, 89, 94],
    [60, 82, 70, 68],
    [88, 84, 85, 90]
])
subjects = ['Math', 'Science', 'English', 'History']

print("Scores Matrix (Students x Subjects):\n", scores)

# Step 2: Calculate each student's average score (mean along columns, axis=1)
student_averages = scores.mean(axis=1)
for i, avg in enumerate(student_averages, 1):
    print(f"Student {i} Average: {avg:.2f}")

# Step 3: Subject with highest class average (mean along rows, axis=0)
subject_averages = scores.mean(axis=0)
best_subject_idx = np.argmax(subject_averages)
print(f"\nHighest scoring subject: {subjects[best_subject_idx]} (Avg: {subject_averages[best_subject_idx]:.2f})")

# Step 4: Students scoring >= 80 in ALL subjects
all_above_80_mask = np.all(scores >= 80, axis=1)
top_students = np.where(all_above_80_mask)[0] + 1
print(f"Students scoring >= 80 in all subjects: Student(s) {top_students.tolist()}")

# Step 5: Min-Max normalization (X - min) / (max - min)
min_val = scores.min()
max_val = scores.max()
normalized_scores = (scores - min_val) / (max_val - min_val)

print("\nNormalized Scores [0 to 1]:\n", np.round(normalized_scores, 2))
```

## Project 2: Daily Temperature & Weather Trends

### Objective
Analyze 14 days of recorded temperature data (in degrees Celsius).

### Tasks
1. Create a 1D NumPy array representing daily temperatures over two weeks.
2. Find the largest day-to-day temperature shift using `np.diff`.
3. Compute a 3-day simple moving average to smooth short-term fluctuations.
4. Identify days that were unusually hot (e.g., above the 75th percentile).
5. Cap extreme heat readings at a threshold (e.g., 35°C) using `np.clip`.

```python
import numpy as np

# Step 1: Daily temperatures for 14 consecutive days (°C)
temperatures = np.array([22.5, 24.0, 23.8, 27.5, 31.0, 36.2, 34.8, 29.0, 26.5, 28.0, 33.5, 37.0, 32.0, 30.5])
print("Daily Temperatures (°C):\n", temperatures)

# Step 2: Largest day-to-day temperature shift
daily_changes = np.diff(temperatures)
max_change_idx = np.argmax(np.abs(daily_changes))
print(f"\nDay-to-day changes: {np.round(daily_changes, 2)}")
print(f"Largest shift occurred from Day {max_change_idx + 1} to Day {max_change_idx + 2}: "
      f"{daily_changes[max_change_idx]:+.1f}°C")

# Step 3: 3-day simple moving average using convolution
window = 3
weights = np.ones(window) / window
moving_avg = np.convolve(temperatures, weights, mode='valid')
print(f"\n3-Day Moving Averages (starting Day 3):\n{np.round(moving_avg, 2)}")

# Step 4: Days above the 75th percentile
p75 = np.percentile(temperatures, 75)
hot_days = np.where(temperatures > p75)[0] + 1
print(f"\n75th Percentile Threshold: {p75:.2f}°C")
print(f"Days above threshold: {hot_days.tolist()} with temps {temperatures[hot_days - 1]}")

# Step 5: Cap temperatures at a maximum safety threshold (35°C)
clipped_temperatures = np.clip(temperatures, a_min=None, a_max=35.0)
print(f"\nTemperatures capped at 35°C:\n{clipped_temperatures}")
```
