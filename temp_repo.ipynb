**Question 1: Basic Single Inheritance**
Create a base class `Animal` with a method `make_sound()` that prints "Some generic sound". Create a derived class `Dog` that inherits from `Animal` and adds a new method `wag_tail()` that prints "Tail wagging". Instantiate a `Dog` object and call both methods.

```python
class Animal:
    def make_sound(self):
        print("Some generic sound")

class Dog(Animal):
    def wag_tail(self):
        print("Tail wagging")

# Solution
my_dog = Dog()
my_dog.make_sound()  # Inherited method
my_dog.wag_tail()    # Dog's own method
```

**Question 2: Using `super()` and Initialization**
Create a base class `Person` that takes `name` and `age` as arguments in its constructor. Create a derived class `Student` that takes `name`, `age`, and `student_id`. Use the `super()` function inside the `Student` constructor to initialize `name` and `age` from the parent class.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, student_id):
        # Call the parent class constructor
        super().__init__(name, age)
        self.student_id = student_id

# Solution
student = Student("Alice", 20, "S12345")
print(f"{student.name} is {student.age} years old with ID {student.student_id}.")
```

**Question 3: Method Overriding**
Create a base class `Vehicle` with a method `get_max_speed()` that returns `100`. Create a subclass `SportsCar` that overrides the `get_max_speed()` method to return `250`. Print the maximum speeds of both a standard `Vehicle` and a `SportsCar`.

```python
class Vehicle:
    def get_max_speed(self):
        return 100

class SportsCar(Vehicle):
    # Overriding the parent method
    def get_max_speed(self):
        return 250

# Solution
standard_vehicle = Vehicle()
fast_car = SportsCar()

print("Vehicle max speed:", standard_vehicle.get_max_speed())
print("SportsCar max speed:", fast_car.get_max_speed())
```

**Question 4: Multiple Inheritance**
Create two separate base classes: `Engine` (with a method `start_engine()`) and `Wheels` (with a method `roll()`). Create a derived class `Motorcycle` that inherits from *both* `Engine` and `Wheels`. Instantiate `Motorcycle` and call both inherited methods.

```python
class Engine:
    def start_engine(self):
        print("Engine started: Vroom!")

class Wheels:
    def roll(self):
        print("Wheels are rolling.")

# Inheriting from multiple classes
class Motorcycle(Engine, Wheels):
    pass 

# Solution
my_bike = Motorcycle()
my_bike.start_engine()
my_bike.roll()
```

**Question 5: Multilevel Inheritance**
Define a class `Employee` with a method `work()`. Define a subclass `Developer` that inherits from `Employee` and adds a method `write_code()`. Finally, define a subclass `SeniorDeveloper` that inherits from `Developer` and adds a method `architect_system()`. Call all three methods from a `SeniorDeveloper` instance.

```python
class Employee:
    def work(self):
        print("Doing general employee tasks.")

class Developer(Employee):
    def write_code(self):
        print("Writing application code.")

class SeniorDeveloper(Developer):
    def architect_system(self):
        print("Designing the system architecture.")

# Solution
senior = SeniorDeveloper()
senior.work()             # Inherited from Employee
senior.write_code()       # Inherited from Developer
senior.architect_system() # Own method
```
