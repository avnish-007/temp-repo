# 20 NumPy Practice Questions & Solutions
Run each cell to verify the output.

```python
import numpy as np
```

### 1. Create a 1D array of integers from 10 to 29.

```python
q1 = np.arange(10, 30)
print(q1)
```

### 2. Create a 3x3 boolean array with all `True` values.

```python
q2 = np.ones((3, 3), dtype=bool)
print(q2)
```

### 3. Extract all odd numbers from an array `[0, 1, 2, ..., 9]`.

```python
arr = np.arange(10)
q3 = arr[arr % 2 != 0]
print(q3)
```

### 4. Replace all odd numbers in `np.arange(10)` with `-1` without modifying the original array.

```python
arr = np.arange(10)
q4 = np.where(arr % 2 != 0, -1, arr)
print(q4)
```

### 5. Reshape a 1D array of 12 elements into a 3x4 matrix.

```python
q5 = np.arange(12).reshape(3, 4)
print(q5)
```

### 6. Stack two arrays `a = np.arange(5)` and `b = np.arange(5, 10)` vertically.

```python
a = np.arange(5)
b = np.arange(5, 10)
q6 = np.vstack([a, b])
print(q6)
```

### 7. Find common elements between two arrays `a = [1, 2, 3, 2, 3, 4]` and `b = [2, 3, 5]`.

```python
a = np.array([1, 2, 3, 2, 3, 4])
b = np.array([2, 3, 5])
q7 = np.intersect1d(a, b)
print(q7)
```

### 8. Get positions where elements of two arrays `a` and `b` match.

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([1, 3, 2, 4, 5])
q8 = np.where(a == b)[0]
print(q8)
```

### 9. Create a 5x5 identity matrix.

```python
q9 = np.eye(5)
print(q9)
```

### 10. Generate 5 evenly spaced numbers between 0 and 1.

```python
q10 = np.linspace(0, 1, 5)
print(q10)
```

### 11. Normalize a 1D random array so its values lie between 0 and 1.

```python
arr = np.array([10, 20, 30, 40, 50])
q11 = (arr - arr.min()) / (arr.max() - arr.min())
print(q11)
```

### 12. Reverse a 1D array.

```python
arr = np.arange(10)
q12 = arr[::-1]
print(q12)
```

### 13. Swap columns 0 and 1 of a 2D array.

```python
arr = np.arange(9).reshape(3, 3)
q13 = arr[:, [1, 0, 2]]
print(q13)
```

### 14. Find the index of the maximum value in an array.

```python
arr = np.array([12, 45, 7, 89, 34])
q14 = np.argmax(arr)
print("Index:", q14, "| Value:", arr[q14])
```

### 15. Check if any element in an array is NaN.

```python
arr = np.array([1, 2, np.nan, 4])
q15 = np.isnan(arr).any()
print(q15)
```

### 16. Compute row-wise and column-wise sums of a 3x3 matrix.

```python
arr = np.arange(1, 10).reshape(3, 3)
row_sum = arr.sum(axis=1)
col_sum = arr.sum(axis=0)
print("Row sums:", row_sum, "| Column sums:", col_sum)
```

### 17. Subtract the row mean from each row of a 2D matrix (broadcasting).

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])
q17 = arr - arr.mean(axis=1, keepdims=True)
print(q17)
```

### 18. Compute the matrix dot product of two 2x2 matrices.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
q18 = a @ b
print(q18)
```

### 19. Count frequencies of unique values in an array.

```python
arr = np.array([1, 2, 2, 3, 3, 3, 4])
vals, counts = np.unique(arr, return_counts=True)
print(dict(zip(vals, counts)))
```

### 20. Clip all values in an array below 10 to 10, and above 30 to 30.

```python
arr = np.array([5, 12, 25, 38, 42])
q20 = np.clip(arr, 10, 30)
print(q20)
```
