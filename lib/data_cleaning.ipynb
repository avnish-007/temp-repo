```python
df['price'] = df['price'].replace("Call for Price",np.nan)
df
```

```python
def clean_price(x):
  x = str(x).lower()
  if "crore" in x:
    return float(x.replace("crore", "").replace("₹","")) * 10000000
  elif "lac" in x or "lakh" in x:
    return (
        float(x.replace("lacs", "").replace("lakh", "").replace("lac", "").replace("₹",""))
        * 100000
    )
  else:
    return None

df['New_price'] = df['price'].apply(clean_price)
df
```
