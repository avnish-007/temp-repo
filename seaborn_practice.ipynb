# Seaborn
It is mainly used for complex data.

### matplotlib: 
- It is used to plot simple data.
- It has limited plots.
- It has limited attributes.

### seaborn:
- It is used for complex data.
- It has more number of plots.
- It has more number of attributes.

### Types of plots:
1. Distribution plot
2. Categorical plot
3. Matrix plot
4. Regression plot

### Types of plots based on number of columns used in plot:
1. Uni-variate plot   : It allows us to plot the graph based on one column.
2. Bi-variate plot    : It allows us to plot the graph based on two columns.
3. Multi-variate plot : It allows us to plot the graph based on multiple columns.

```python
# 1. Distribution plot:
#                   It help us to understand how the data got distributed

# --> Distribution -- Spread of Values.
# By this what can we analyse?
# -> min, max, highly dense values, unique values.

# Types of distribution plot:
# 1. histplot()
# 2. jointplot()
# 3. pairplot()
```

```python
# histplot():
#           It is used to represent count the unique values.

# Import required visualization and data manipulation libraries
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Tips dataset
Tips = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/tips.csv")
Tips
```

```python
# Syntax:
#       sns.histplot(x='col_name',data=var)

# Univariate histplot with 'hue' for category grouping (smoker vs non-smoker)
a = sns.histplot(x='sex',data=Tips,hue='smoker',palette='Oranges')     #palette= 'inferno'/'viridis'/'magma'/'Set1'/'Set2'/'Set3'

# Annotate bar containers with exact count labels
for i in a.containers:
    a.bar_label(i)
```

```python
# Load Wine Quality dataset
WineQuality = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/winequalityN.csv")
WineQuality
```

```python
# Plot wine quality distribution grouped by wine type with custom weights and 5 bins
sns.histplot(x='quality',data=WineQuality,weights=5,hue='type',palette='Reds',bins=5)
# Display plot cleanly
plt.show()
```

```python
# Load HR Analytics dataset
hr_data = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/hr_data.csv")
hr_data
```

```python
# Plot salary category distribution with distinct hue colors per salary group
b = sns.histplot(x='salary',data=hr_data,hue='salary')

# Add count labels on top of each bar
for i in b.containers:
    b.bar_label(i)
```

```python
# jointplot():

# Syntax:
#       sns.jointplot(x='col_name',y='col_name',data=var)

sns.jointplot(x='total_bill',y='tip',data=Tips, hue='sex')
```

```python
sns.jointplot(x='pH',y='density',data=WineQuality,hue='type')
```

```python
# pairplot():
#           It is used to understand the relationship between each column.
# Syntax:
#       sns.pairplot(var)

sns.pairplot(Tips)
```

```python
# It is used to analyse a data based on catagorical columns.
# 1. countplot
# 2. barplot
# 3. boxplot
# 4. scatterplot
# 5. violinplot
```

```python
# countplot()
#           It is used to count the unique values.
# Syntax:
#       sns.countplot(x='col_name',data=var)

sm = sns.countplot(x='smoker',data=Tips,hue='sex')
for i in sm.containers:
    sm.bar_label(i)
```

```python
titanic = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/titanic.csv")
titanic
```

```python
mf = sns.countplot(x='sex',data=titanic)
for i in mf.containers:
    mf.bar_label(i)
```

```python
wine = sns.countplot(x="type",data=WineQuality)
for i in wine.containers:
    wine.bar_label(i)
```

```python
# barplot():
#           It is used to know the average/mean value.
# Syntax:
#       sns.barplot(x='col_name',y="col_name",data=var)

quality = sns.barplot(x='type',y='quality',data=WineQuality,errorbar=None)
for i in quality.containers:
    quality.bar_label(i)
```

```python
avg_age = sns.barplot(x="age", y='sex',data=titanic,errorbar=None,hue='sex')
for i in avg_age.containers:
    avg_age.bar_label(i)
```
