# Seaborn
It is mainly used for complex data.

### matplotlib: 
- It is used to plot simple data.
- It has limited plots.
- It has limited attributes.

### seaborn:
- It is used for complex data.
- It has more number of plots.
- It has more number of attributes.

### Types of plots:
1. Distribution plot
2. Catogrical plot
3. Matrix plot
4. Regression plot

### Types of plots based on number of columns used in plot:
1. Uni-variate plot   : It allows us to plot the graph based on one column.
2. Bi-variate plot    : It allows us to plot the graph based on two columns.
3. Multi-variate plot : It allows us to plot the graph based on multiple columns.

```python
# 1. Distribution plot:
#                   It help us to understand how the data got distributed

# --> Distribution -- S;pread of Values.
# By this what can we analyse?
# -> min, max, highly dense values, unique values.

# Types of distribution plot:
# 1. histplot()
# 2. jointplot()
# 3. pairplot()
```

```python
# histplot():
#           It is used to represent count the unique values.

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Tips = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/tips.csv")
Tips
```

```python
# Syntax:
#       sns.histplot(x='col_name',data=var)

a = sns.histplot(x='sex',data=Tips,hue='smoker',palette='Oranges')     #palette= 'inferno'/'viridis'/'magma'/'Set1'/'Set2'/'Set3'
for i in a.containers:
    a.bar_label(i)
```

```python
WineQuality = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/winequalityN.csv")
WineQuality
```

```python
sns.histplot(x='quality',data=WineQuality,weights=5,hue='type',palette='Reds',bins=5)
plt.show()
```
