# Seaborn
It is mainly used for complex data.

### matplotlib: 
- It is used to plot simple data.
- It has limited plots.
- It has limited attributes.

### seaborn:
- It is used for complex data.
- It has more number of plots.
- It has more number of attributes.

### Types of plots:
1. Distribution plot
2. Categorical plot
3. Matrix plot
4. Regression plot

### Types of plots based on number of columns used in plot:
1. Uni-variate plot   : It allows us to plot the graph based on one column.
2. Bi-variate plot    : It allows us to plot the graph based on two columns.
3. Multi-variate plot : It allows us to plot the graph based on multiple columns.

```python
# 1. Distribution plot:
#                   It help us to understand how the data got distributed

# --> Distribution -- Spread of Values.
# By this what can we analyse?
# -> min, max, highly dense values, unique values.

# Types of distribution plot:
# 1. histplot()
# 2. jointplot()
# 3. pairplot()
```

```python
# histplot():
#           It is used to represent count the unique values.

# Import required visualization and data manipulation libraries
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Tips dataset
Tips = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/tips.csv")
Tips
```

```python
# Syntax:
#       sns.histplot(x='col_name',data=var)

# Univariate histplot with 'hue' for category grouping (smoker vs non-smoker)
a = sns.histplot(x='sex',data=Tips,hue='smoker',palette='Oranges')     #palette= 'inferno'/'viridis'/'magma'/'Set1'/'Set2'/'Set3'

# Annotate bar containers with exact count labels
for i in a.containers:
    a.bar_label(i)
```

```python
# Load Wine Quality dataset
WineQuality = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/winequalityN.csv")
WineQuality
```

```python
# Plot wine quality distribution grouped by wine type with custom weights and 5 bins
sns.histplot(x='quality',data=WineQuality,weights=5,hue='type',palette='Reds',bins=5)
# Display plot cleanly
plt.show()
```

```python

# boxplot():
#           It is used to identify outliers.
# Syntax:
#       sns.boxplot(x='col_name',y='col_name',data=var_name)

sns.boxplot(x='type',y='pH',data=WineQuality)
```

```python
# To handle Outliers we'll use IQR (Inter Quartile Range) method.
# formula: q3 - q1
q3 = WineQuality['pH'].quantile(0.75)
q1 = WineQuality['pH'].quantile(0.25)
ans = q3 - q1
ans
```

```python
lb = q1 - 0.5 * ans
ub = q3 + 0.5 * ans
display([lb,ub])
```

```python
ol_wine = []
for i in WineQuality['pH']:
    if i > ub:
        ol_wine.append(ub)
    elif i < lb:
        ol_wine.append(lb)
    else:
        ol_wine.append(i)
WineQuality['pH'] = ol_wine
```

```python
sns.boxplot(y='pH',x='type',data=WineQuality)
```

```python
sns.boxplot(x='sex',y='age',data=titanic)
```

```python
# To handle Outliers we'll use IQR (Inter Quartile Range) method.
# formula: q3 - q1
q3 = titanic['age'].quantile(0.75)
q1 = titanic['age'].quantile(0.25)
ans = q3 - q1
ans
```

```python
lb = q1 - 0.5*ans
ub = q3 + 0.5*ans
display([lb,ub])
```

```python
ol = titanic[(titanic['age'] < ub) & (titanic['age'] > lb)]
ol
```

```python
sns.boxplot(x='sex',y='age',data=ol)
```

```python
# To manage outliers
out = []
for i in titanic['age']:
    if i > ub:
        out.append(ub)
    elif i < lb:
        out.append(lb)
    else:
        out.append(i)

titanic['age'] = out
```

```python
sns.boxplot(x='sex',y='age',data=titanic)
```

```python
# Scatterplot():
#               It is used to identify the relation between 2 columns.
# Syntax:
#       sns.scatterplot(x='col_1','y='col_2',data=var)

sns.scatterplot(y='pH',x='type',data=WineQuality)
```

```python
sns.violinplot(x='type',y='pH',data=WineQuality)
```

```python
sns.violinplot(x='sex',y='age',data=titanic,split=True)
```

```python
# 
# It is used to compare one column with respect to another column.
# It is mainly used in case of machine learning.
# It is used to observe relation between multiple columns.
# Syntax:
#       var = ['col_1','col_2',. . ., 'col_n']
#       sns.heatmap(data=var[var_1].corr(),annote=True, cmap='plalette_color')

cols = ['fixed acidity','volatile acidity','chlorides','density','sulphates','pH']
sns.heatmap(WineQuality[cols].corr(),annot=True)
```
