```python
# Import pandas library for data manipulation
import pandas as pd

# Load the Amazon forest fires dataset with ISO-8859-1 encoding to support Portuguese characters
amazon_fires = pd.read_csv("https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/amazon_fires.csv",encoding='ISO-8859-1')
amazon_fires
```

```python
# Rename Portuguese column names to English equivalents for better readability
# 'ano' -> 'year', 'mes' -> 'month', 'estado' -> 'state', 'numero' -> 'no_of_fires', 'encontro' -> 'date'
amazon_fires.rename(columns={'ano':'year','mes':'month','estado':'state','numero':'no_of_fires','encontro':'date'},inplace=True)
amazon_fires
```

```python
# Check unique month values present in the dataset (currently in Portuguese)
amazon_fires['month'].unique()
```

```python
# Dictionary mapping Portuguese month names to abbreviated English month names
amz = {'Janeiro':'Jan','Fevereiro':'Feb','Maro':'Mar','Abril':'Apr','Maio':'May','Junho':'Jun','Julho':'July','Agosto':'Aug','Setembro':'Sep','Outubro':'Oct','Novembro':'Nov','Dezembro':'Dec'}
```

```python
# Map and replace Portuguese month names with English month abbreviations
amazon_fires['month'] = amazon_fires['month'].map(amz)
amazon_fires
```

```python
# Strip out the string suffix 'Fires' from 'no_of_fires' to isolate numeric values
amazon_fires['no_of_fires'] = amazon_fires['no_of_fires'].str.strip('Fires')
```

```python
# Convert 'no_of_fires' from string data type to float for numerical analysis
amazon_fires['no_of_fires'] = amazon_fires['no_of_fires'].astype('float')
amazon_fires
```

```python
# Calculate median of 'no_of_fires' column
med = amazon_fires['no_of_fires'].median()

# Impute missing (NaN) values in 'no_of_fires' using the calculated median
amazon_fires['no_of_fires'].fillna(med)
amazon_fires
```
