```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("D:/New folder/temp_repo/data/anime.csv")
```

```python
df.loc[2]["Title"]
```

```python
def extract_episodes(txt):
    check = False
    ep = ""
    for i in txt:
        if i == ")":
            break
            
        if check == True:
            ep += i
        if i == "(":
            check = True
    return ep
```

```python

df["episodes"] = df["Title"].apply(extract_episodes)
```

```python
df["episodes"] = df["episodes"].str.replace(" eps","")
```

```python

df["episodes"] = df["episodes"].astype(int)
```

```python
df.loc[2]["Title"]
```

```python
def extracting_timePeriod(txt):
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                data += txt[j]
            return data
```

```python
df["Total Time"] = df["Title"].apply(extracting_timePeriod)
```
